```python
coor = []

map = folium.Map(location=centro, zoom_start=12, tiles='https://server.arcgisonline.com/ArcGIS/rest/services/World_Street_Map/MapServer/tile/{z}/{y}/{x}', attr='Esri')

for _, row in cafe.iterrows():
    lat = row['Latitud']
    lon = row['Longitud']
    coor.append([lat, lon])

    content_popup = f"""
    <div style='width:200px'>
        <h4>{row['Nombre']}</h4>
        <b>Dirección:</b> {row['Direccion']}<br>
        <b>Calificación:</b> ⭐ {row['Calificacion']}<br>
        <b>Rango de Precios:</b> {row['Precio_Min_PYG']:,} - {row['Precio_Max_PYG']:,} PYG
    </div>
    """

    folium.Marker(
        location=[lat, lon],
        popup=folium.Popup(content_popup, max_width=250),
        tooltip=row['Nombre'],
        icon=folium.Icon(color='orange', icon='star')
    ).add_to(map)
```
